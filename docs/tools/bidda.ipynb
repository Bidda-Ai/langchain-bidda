# BiddaComplianceTool

[Bidda](https://bidda.com) is a registry of source-verified regulatory and compliance intelligence: 10,000+ obligations across 39 regulated pillars, each traced to a primary legal source with a content hash.

This notebook shows how to use the `BiddaComplianceTool` so a LangChain agent can look up what a regulation requires and cite a primary source, instead of relying on the model's own recall. It is reference intelligence, not legal advice; the reader draws the legal conclusion.

### Tool features

The tool searches the free discovery tier of the Bidda API, so no API key is required. Full node payloads (deterministic workflow, actionable schema, full citation list) are available to authenticated callers via the Bidda API.

### Credentials

No credentials are required for discovery-tier search. If you also want to pull full (gated) node payloads, pass a `skyfire_token` or `base_tx_hash` when constructing the tool.

## Instantiation

Here we show how to instantiate an instance of the tool:

```python
from langchain_bidda import BiddaComplianceTool

tool = BiddaComplianceTool(
    # pillar="ai-governance",  # optional: restrict to one pillar slug
    # max_results=4,           # optional: related node ids to surface
)
```

## Invocation

### Invoke directly with args

```python
print(tool.invoke({"query": "EU AI Act Article 10"}))
```

### Invoke with ToolCall

We can also invoke the tool with a model-generated `ToolCall`, in which case a `ToolMessage` will be returned:

```python
model_generated_tool_call = {
    "args": {"query": "GDPR breach notification deadline"},
    "id": "1",
    "name": tool.name,
    "type": "tool_call",
}
tool.invoke(model_generated_tool_call)
```

## Use within an agent

Bind the tool to a tool-calling model and let the agent decide when to look up an obligation:

```python
# pip install -qU "langchain[openai]"
from langchain.chat_models import init_chat_model
from langchain_bidda import BiddaComplianceTool

llm = init_chat_model("gpt-4o-mini", model_provider="openai")
llm_with_tools = llm.bind_tools([BiddaComplianceTool()])

llm_with_tools.invoke(
    "What does the EU AI Act require for training-data governance? Cite the source."
)
```
